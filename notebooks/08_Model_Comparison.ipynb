```python
# Import pandas for creating and saving the results table
import pandas as pd

# Import matplotlib for creating comparison graphs
import matplotlib.pyplot as plt

# Import os for working with folders and file paths
import os
```

```python
# Define the path to the results folder
results_folder = "../results"

# Create the results folder if it does not already exist
os.makedirs(results_folder, exist_ok=True)

# Display the folder path
print("Results folder:", results_folder)
```

```python
# Evaluate saved artifacts on the common holdout split.
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))
from digit_recognition.evaluation import evaluate_saved_models
results_percent = evaluate_saved_models()
```

```python
# Metrics are already expressed as percentages.
results_percent
```

```python
# Save the comparison table as a CSV file
results_percent.to_csv(
    "../results/model_comparison.csv",
    index=False
)

# Confirm that the file was saved
print("Model comparison saved successfully!")
```

```python
# Find the row containing the highest accuracy
best_model = results_percent.loc[
    results_percent["Accuracy"].idxmax()
]

# Display the best model
print("Best Algorithm:", best_model["Algorithm"])
print("Best Accuracy:", best_model["Accuracy"], "%")
```

```python
def plot_comparison(metric, title, filename):
    ranked = results_percent.sort_values(metric, ascending=metric != "Error Rate")
    best = ranked[metric].idxmin() if metric == "Error Rate" else ranked[metric].idxmax()
    colors = ["#28634A" if index == best else "#AABDAF" for index in ranked.index]
    fig, ax = plt.subplots(figsize=(8, 4.4))
    bars = ax.barh(ranked["Algorithm"], ranked[metric], color=colors, height=0.6)
    ax.bar_label(bars, labels=[f"{value:.2f}%" for value in ranked[metric]], padding=5, fontsize=10)
    ax.set_title(title, loc="left", fontsize=15, pad=16, fontweight="bold")
    ax.set_xlabel(f"{metric} (%)")
    ax.set_xlim(0, max(10, ranked[metric].max() * 1.2) if metric == "Error Rate" else 108)
    ax.set_axisbelow(True)
    ax.grid(axis="x", alpha=0.18)
    ax.tick_params(axis="both", length=0)
    for spine in ax.spines.values():
        spine.set_visible(False)
    fig.tight_layout()
    fig.savefig(f"{results_folder}/{filename}", dpi=180, bbox_inches="tight")
    plt.show()
    plt.close(fig)

plot_comparison("Accuracy", "Holdout accuracy", "accuracy_comparison.png")
```

```python
plot_comparison("Precision", "Weighted precision", "precision_comparison.png")
```

```python
plot_comparison("Recall", "Weighted recall", "recall_comparison.png")
```

```python
plot_comparison("F1-Score", "Weighted F1 score", "f1_comparison.png")
```

```python
plot_comparison("Error Rate", "Holdout error rate", "error_rate_comparison.png")
```
