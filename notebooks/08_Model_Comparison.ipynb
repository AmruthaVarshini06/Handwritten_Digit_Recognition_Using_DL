```python
# Import pandas for creating and saving the results table
import pandas as pd

# Import matplotlib for creating comparison graphs
import matplotlib.pyplot as plt

# Import os for working with folders and file paths
import os
```

```python
# Define the path to the results folder
results_folder = "../results"

# Create the results folder if it does not already exist
os.makedirs(results_folder, exist_ok=True)

# Display the folder path
print("Results folder:", results_folder)
```

```python
# Create a dictionary containing the results
# obtained from each machine learning algorithm

results = {
    "Algorithm": [
        "MLP",
        "SVM",
        "KNN",
        "Random Forest",
        "Logistic Regression",
        "CNN"
    ],

    "Accuracy": [
        0.9742857142857143,
        0.9791666666666666,
        0.9677380952380952,
        0.9639285714285715,
        0.9128571428571428,
        0.9830952867220276
    ],

    "Precision": [
        0.974293,
        0.9791557,
        0.968166,
        0.9639276,
        0.912623,
        0.983494370826049
    ],

    "Recall": [
        0.9742857142857143,
        0.9791666666666666,
        0.9677380952380952,
        0.9639285714285715,
        0.9128571428571428,
        0.9830952867220276
    ],

    "F1-Score": [
        0.974285,
        0.9791527,
        0.9676708,
        0.9639004,
        0.9126200,
        0.9831390274608462
    ],

    "Error Rate": [
        0.0257142857142857,
        0.0208333333333333,
        0.0322619047619047,
        0.0360714285714285,
        0.0871428571428572,
        0.0169047132779724
    ]
}
```

```python
# Convert the results dictionary into a pandas DataFrame
results_df = pd.DataFrame(results)

# Display the complete comparison table
results_df
```

```python
# Create a copy of the results DataFrame
results_percent = results_df.copy()

# Convert Accuracy to percentage
results_percent["Accuracy"] = results_percent["Accuracy"] * 100

# Convert Precision to percentage
results_percent["Precision"] = results_percent["Precision"] * 100

# Convert Recall to percentage
results_percent["Recall"] = results_percent["Recall"] * 100

# Convert F1-Score to percentage
results_percent["F1-Score"] = results_percent["F1-Score"] * 100

# Convert Error Rate to percentage
results_percent["Error Rate"] = results_percent["Error Rate"] * 100

# Round the values to 2 decimal places
results_percent = results_percent.round(2)

# Display the results
results_percent
```

```python
# Save the comparison table as a CSV file
results_percent.to_csv(
    "../results/model_comparison.csv",
    index=False
)

# Confirm that the file was saved
print("Model comparison saved successfully!")
```

```python
# Find the row containing the highest accuracy
best_model = results_percent.loc[
    results_percent["Accuracy"].idxmax()
]

# Display the best model
print("Best Algorithm:", best_model["Algorithm"])
print("Best Accuracy:", best_model["Accuracy"], "%")
```

```python
# Create a new figure
plt.figure(figsize=(4,4))

# Create a bar chart for accuracy
plt.bar(
    results_percent["Algorithm"],
    results_percent["Accuracy"]
)

# Add the title
plt.title("Accuracy Comparison of Machine Learning Algorithms")

# Label the x-axis
plt.xlabel("Algorithm")

# Label the y-axis
plt.ylabel("Accuracy (%)")

# Rotate algorithm names for better readability
plt.xticks(rotation=55)

# Add a grid to the y-axis
plt.grid(axis="y", alpha=0.1)

# Adjust the layout
plt.tight_layout()

# Save the graph
plt.savefig(
    "../results/accuracy_comparison.png",
    dpi=200,
    bbox_inches="tight"
)

# Display the graph
plt.show()
```
